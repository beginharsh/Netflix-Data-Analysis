```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("netflix_titles.csv")
```

```python
df.sample(5)
```

```python
df.shape
```

```python
df.size
```

```python
df.index
```

```python
df.columns
```

```python
df.drop(['show_id','description'],axis=1,inplace=True)
```

```python
df.sample(5)
```

```python
df.isna().sum() # Checking count of null values in each Column
```

```python
df["country"].value_counts().head(3) # Top 3 highest occuring countries
```

```python
mode_country = df["country"].mode()[0] # Highest Occuring State
mode_country
```

```python
df["country"] = df["country"].fillna(mode_country) # replacing country column null value to mode value of that column.
```

```python
df["country"].isna().sum() # cross checking to make sure that changes happened.
```

```python
mode_date = df["date_added"].mode()[0]
mode_date
```

```python
df["date_added"] = df["date_added"].fillna(mode_date)
```

```python
df["date_added"].isna().sum()
```

```python
mode_rating = df["rating"].mode()[0]
mode_rating
```

```python
df["rating"] = df["rating"].fillna(mode_rating)
```

```python
df["rating"].isna().sum()
```

```python
mode_duration = df["duration"].mode()[0]
mode_duration
```

```python
df["duration"] = df["duration"].fillna(mode_duration)
df["duration"].isna().sum()
```

```python
df["director"] = df["director"].fillna("Not Mentioned") # Replacing null values with Default Value -- Not Mentioned
```

```python
df["director"].isna().sum()
```

```python
df["cast"] = df["cast"].fillna("Not Mentioned") # Relacing Null Values with Default Value
df["cast"].isna().sum()
```

```python
df.isna().sum()
```

```python
df["rating"].unique() # checking the unique/distinct values in rating column
```

```python
df = df[~df["rating"].str.contains("min")] # removing the wrong format rating rows
df["rating"].unique()                                 
```

```python
df["date_added"].unique()
```

```python
df["date_added"] = df["date_added"].str.replace(",","") # replacing/removing the comma with blank in date_added Column 
df["date_added"].unique()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"],format="mixed") # function to convert strings to datetime objects
df["date_added"].unique()
```

```python
pd.isnull(df).sum()
```

```python
df["Year"] = df["date_added"].dt.year # extracting year from a datetime column
```

```python
df["Month"] = df["date_added"].dt.month # extracting month from a datetime column
```

```python
df["Day"] = df["date_added"].dt.day # extracting day from a datetime column
```

```python
df.sample(5)
```

```python
df["Genre"] = df["listed_in"].str.split(",").str[0] # splitting the column using comma(,) and making a new column
```

```python
df["Genre"].value_counts().head(5)
```

```python
df.sample(5)
```

```python
df = df[df["cast"] != "Not Mentioned"] # Removing rows where cast = Not Mentioned
df["cast"].value_counts()
```

```python
df["Lead_actor"] = df["cast"].str.split(",").str[0] # Splitting the column by , and making a new column - Lead_actor
df["Lead_actor"].value_counts().head(5)
```

```python
df.sample(5)
```

```python
df.drop(["cast","date_added","listed_in"], axis=1, inplace=True) # Dropping the extra columns 
```

```python
df.sample(5)
```

```python
df.to_csv("Transformed_data.csv")
```
